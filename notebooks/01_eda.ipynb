```python
import sys
sys.path.append('..')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.utils import load_data, set_seed
from src.metric import compute_competition_metric

set_seed(42)
plt.style.use('seaborn-v0_8-whitegrid')
%matplotlib inline
```

## 1. Load Data

```python
train_df, test_df, sample_sub = load_data()
print(f"Train shape: {train_df.shape}")
print(f"Test shape:  {test_df.shape}")
train_df.head()
```

## 2. Label Distribution Analysis
Analyzing the reading preferences (1 = do not read, 2-5 = reading priority)

```python
label_counts = train_df['Label'].value_counts().sort_index()
print("Label counts:")
print(label_counts)
print("\nPercentage:")
print(label_counts / len(train_df) * 100)

plt.figure(figsize=(8, 4))
sns.barplot(x=label_counts.index, y=label_counts.values, palette='viridis')
plt.title('Distribution of Reading Preference Ratings (1 to 5)')
plt.xlabel('Label')
plt.ylabel('Number of Papers')
plt.show()
```

## 3. Publication Year Analysis

```python
plt.figure(figsize=(10, 4))
sns.histplot(data=train_df, x='year', hue='Label', multiple='stack', bins=30, palette='magma')
plt.title('Paper Publication Year by Label')
plt.xlabel('Year')
plt.ylabel('Count')
plt.show()
```

## 4. Title Text & Topic Exploration

```python
train_df['title_len'] = train_df['title'].fillna('').apply(len)
train_df['word_count'] = train_df['title'].fillna('').apply(lambda s: len(s.split()))

print("Average title length (characters):", train_df['title_len'].mean())
print("Average title word count:", train_df['word_count'].mean())

fig, ax = plt.subplots(1, 2, figsize=(14, 4))
sns.boxplot(data=train_df, x='Label', y='word_count', ax=ax[0], palette='Blues')
ax[0].set_title('Word Count vs Label')

sns.boxplot(data=train_df, x='Label', y='title_len', ax=ax[1], palette='Greens')
ax[1].set_title('Character Length vs Label')
plt.show()
```
